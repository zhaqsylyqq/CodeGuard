# 03. BiLSTM

Словарь строится только по train. Две LSTM читают токены в обоих направлениях. Лучший checkpoint выбираем по val loss; test не используем. Перед запуском установите requirements-neural.txt в окружение ядра.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
assert (ROOT / "src").is_dir(), "Open the notebook from the CodeGuard repository"

def run_script(name, *args):
    subprocess.run([sys.executable, str(ROOT / "src" / name), *map(str, args)], cwd=ROOT, check=True)
```

```python
run_script("bilstm.py", "--epochs", 10, "--output", "runs/bilstm")
```

```python
import json
metrics = json.loads((ROOT / "runs/bilstm/metrics.json").read_text(encoding="utf-8"))
{k: metrics[k] for k in ("accuracy", "precision", "recall", "f1", "roc_auc", "best_epoch")}
```
